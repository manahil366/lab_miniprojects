### D7. One method calls another

1. Copy your class from D6 and add a method `percent(self)` that returns `(self.value / self.limit) * 100`.
2. Change `status(self)` so it uses `self.percent()` instead of working out the percentage itself.
3. Create `Record("R-001", 87.0, 100.0)` and print its `.percent()` and `.status()`.

**Expected output:** `87.0`, then `OK`

````python
# D7
class Record:
    def __init__(self, label, value, limit):
        self.label = label
        self.value = value
        self.limit = limit

    def percent(self):
        return (self.value/self.limit) * 100        #defined the percent function seperately

    def status(self):
        p = self.percent()                          #define a variable and use it for comparisons

        if p >= 100:
            return "OVERLIMIT"

        elif p >= 90:
            return "WARNING"

        else:
            return "OK"

R5 = Record("R-001",87.0,100.0)
print(R5.percent())
print(R5.status())

        

````

### D8. A method can change its own object

1. Define a class `Checker`. In `__init__(self)`, set `self.over_limit_count = 0`.
2. Add a method `check(self, value, limit)` that adds 1 to `self.over_limit_count` when `value` is
   greater than `limit`.
3. Create one `Checker`, call `.check()` three times with `(87, 100)`, `(120, 100)` and `(130, 100)`,
   then print `over_limit_count`.

**Expected output:** `2`

````python
# D8
class Checker:
    def __init__(self):
        self.over_limit_count = 0

    def check(self,value,limit):
        if value > limit:
            self.over_limit_count += 1

c = Checker()
c.check(87,100)
c.check(120,100)
c.check(130,100)
print(c.over_limit_count)
````

### C2. An account that locks

Write a class `Account` with a `username` and a count of failed logins that starts at 0.
It has two methods:

- `fail(self)`: adds 1 to the failed logins
- `locked(self)`: returns `True` if there have been 3 or more failed logins

**Test it:**

| Code | Expected output |
|---|---|
| an account for `"anna"` with `.fail()` called 3 times, then `print(anna.locked())` | `True` |
| an account for `"ben"` with `.fail()` called once, then `print(ben.locked())` | `False` |

````python
# C2
class Account:
    def __init__(self,username):
        self.username = username     #defined the attribute
        self.failed_logins = 0       #set faile logins count to 0
        

    def fail(self):
        self.failed_logins += 1     #increment the count if detect a failed login attempt

    def locked(self):
        return self.failed_logins >= 3  #will get locked if failed attempts exceed 3

#input given
anna = Account("anna")
anna.fail()
anna.fail()
anna.fail()
print(anna.locked())

ben = Account("ben")
ben.fail()
print(ben.locked())
````

````python
"""
RECORD CHECK  -  my version
===========================

Name  :  MANAHIL SHAHEEN
Lane  :  AI
Date  :  23-10-26


"""

# ====================================================================== CLASS
# 1. Define a class called Record.
#
#    Threshold : __init__(self, label, value, limit) storing all three as
#                attributes. No methods yet.
#    Typical   : add a method status(self) that returns "OVER LIMIT" (100%+),
#                "WARNING" (90%+) or "OK", worked out from self.value and
#                self.limit. Give it a one-line docstring.
#    Excellent : also add over_limit_count as a class-level running total -
#                see the note below.




class Record:
    def __init__(self, dataset_name, rows_loaded, rows_expected):   #THRESHOLD
        """Store the dataset details."""
        self.dataset_name = dataset_name
        self.rows_loaded = rows_loaded
        self.rows_expected = rows_expected              #defined the attributes

    def percent(self):
        """Return the percentage of expected rows loaded."""
        return (self.rows_loaded / self.rows_expected) * 100      #seperate percent method

    def status(self):                                                #TYPICAL
        """Return the status based on the percentage."""
        percent = self.percent()

        if percent >= 100:
            return "OVERLIMIT"
        elif percent >= 90:
            return "WARNING"                               #seperate status method
        else:
            return "OK"


# ==================================================================== RECORDS
# 2. Create your records.
#
#    Threshold : create at least 2 Record instances directly in the code.
#    Typical   : same, but now print each one's .status() as well.
#    Excellent : instead, write load_records(path) that reads label,value,limit
#                lines from a file and returns a list of Record objects. Make
#                your own data file, one record per line.




def load_records(path):                                  #EXCELLENT (OVERWROTE THRESHOLD AND TYPICAL)
    """Read records from a file and return Record objects."""
    records = []

    with open(path, "r") as file:
        for line in file:
            line = line.strip()

            if line == "":
                continue

            dataset_name, rows_loaded, rows_expected = line.split(",")

            record = Record(
                dataset_name,
                int(rows_loaded),
                int(rows_expected)
            )

            records.append(record)

    return records


records = load_records("records_sample_UPDATED.txt")        #loaded the records and even tested with an updated file(REMOVE_UPDATED for normal file)


# ==================================================================== OUTPUT
# 3. Print a report for each record.
#
#    Threshold : label, value, limit, inside a border.
#    Typical   : add the result of .status().
#    Excellent : also keep a running count of how many came back OVER LIMIT
#                across every record, and print that once, after the loop.
#                (Simplest: a plain variable outside the loop, same pattern
#                as Week 4. If you want it to live ON the object instead,
#                that is a stretch worth trying - but a variable is enough.)



over_limit_count = 0              #EXCELLENT FROM STEP 1

for record in records:
    print()
    print("=" * 34)
    print(f"  RECORD CHECK - {record.dataset_name}")
    print("=" * 34)
    print(f"Dataset:       {record.dataset_name}")
    print(f"Rows loaded:   {record.rows_loaded}")
    print(f"Rows expected: {record.rows_expected}")       #THRESHOLD
    print(f"Percentage:    {record.percent():.1f}%")
    print(f"Status:        {record.status()}")                    #TYPICAL
    print("=" * 34)

    if record.status() == "OVERLIMIT":
        over_limit_count += 1                              #EXCELLENT

print(f"\nTotal over-limit records: {over_limit_count}")    #an f string to display the over_limit_count




````
